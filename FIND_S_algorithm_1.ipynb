### FIND-S Algorithm Implementation and Demonstration

```python
import os
import pandas as pd

def find_csv_file():
    for root, dirs, files in os.walk('/content'):
        for file in files:
            if 'enjoysport' in file:
                return os.path.join(root, file)
    for root, dirs, files in os.walk('/content'):
        if 'enjoysport' in root:
            return root
    return None

target_path = find_csv_file()
if target_path:
    # Read without headers since the first row is actual data
    df = pd.read_csv(target_path, header=None)
    print("Successfully loaded dataset from:", target_path)
    display(df)

    # Extract features and targets
    csv_data = df.iloc[:, :-1].values.tolist()
    csv_target = [str(label).strip().capitalize() for label in df.iloc[:, -1].values.tolist()]

    print(f"\n The total number of training instances are :  {len(csv_data)}")

    # Initialize hypothesis
    hypothesis = ['0'] * len(csv_data[0])
    print("\n The initial hypothesis is : ")
    print(hypothesis)

    # Run FIND-S with step-by-step tracing
    first_positive = True
    for idx, (instance, label) in enumerate(zip(csv_data, csv_target), 1):
        if label == 'Yes':
            if first_positive:
                hypothesis = list(instance)
                first_positive = False
            else:
                for i in range(len(hypothesis)):
                    if instance[i] != hypothesis[i]:
                        hypothesis[i] = '?'
        print(f"\n The hypothesis for the training instance {idx} is : ")
        print(hypothesis)

    print("\n The Maximally specific hypothesis for the training instance is ")
    print(hypothesis)
else:
    print("Could not automatically locate the enjoysport dataset file.")
```
